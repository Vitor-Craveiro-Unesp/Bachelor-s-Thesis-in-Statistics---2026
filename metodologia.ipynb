```python
# ============================================================
# RELATÓRIO DO AMBIENTE COMPUTACIONAL
# Python / Jupyter / Bibliotecas / Sistema
# ============================================================

import sys
import platform
import subprocess
import os
import pkg_resources

# ------------------------------------------------------------
# 1. Informações do Python
# ------------------------------------------------------------

print("=" * 80)
print("INFORMAÇÕES DO PYTHON")
print("=" * 80)

print(f"Versão do Python: {sys.version}")
print(f"Executável Python: {sys.executable}")
print(f"Ambiente virtual: {os.environ.get('VIRTUAL_ENV', 'Não identificado')}")

# ------------------------------------------------------------
# 2. Informações do sistema operacional
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("INFORMAÇÕES DO SISTEMA OPERACIONAL")
print("=" * 80)

print(f"Sistema operacional: {platform.system()}")
print(f"Versão do sistema: {platform.version()}")
print(f"Release: {platform.release()}")
print(f"Arquitetura: {platform.architecture()[0]}")
print(f"Nome da máquina: {platform.node()}")
print(f"Processador: {platform.processor()}")

# ------------------------------------------------------------
# 3. Informações de CPU e memória RAM
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CPU E MEMÓRIA RAM")
print("=" * 80)

try:
    import psutil

    print(f"Núcleos físicos da CPU: {psutil.cpu_count(logical=False)}")
    print(f"Núcleos lógicos da CPU: {psutil.cpu_count(logical=True)}")

    memoria = psutil.virtual_memory()
    memoria_total_gb = memoria.total / (1024 ** 3)

    print(f"Memória RAM total: {memoria_total_gb:.2f} GB")

except ImportError:
    print("Biblioteca psutil não instalada.")
    print("Para instalar, execute: pip install psutil")

# ------------------------------------------------------------
# 4. Informações do Jupyter
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("INFORMAÇÕES DO JUPYTER")
print("=" * 80)

try:
    resultado = subprocess.run(
        ["jupyter", "--version"],
        capture_output=True,
        text=True
    )

    print(resultado.stdout)

except Exception as e:
    print("Não foi possível obter informações do Jupyter.")
    print(f"Erro: {e}")

# ------------------------------------------------------------
# 5. Informações do VS Code
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("INFORMAÇÕES DO VS CODE")
print("=" * 80)

try:
    resultado = subprocess.run(
        ["code", "--version"],
        capture_output=True,
        text=True
    )

    if resultado.stdout:
        print(resultado.stdout)
    else:
        print("VS Code não encontrado pelo comando 'code'.")
        print("Talvez o comando 'code' não esteja habilitado no PATH.")

except Exception as e:
    print("Não foi possível obter informações do VS Code.")
    print(f"Erro: {e}")

# ------------------------------------------------------------
# 6. Versões das principais bibliotecas
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("VERSÕES DAS PRINCIPAIS BIBLIOTECAS")
print("=" * 80)

bibliotecas = [
    "pandas",
    "numpy",
    "scikit-learn",
    "matplotlib",
    "seaborn",
    "plotly",
    "scipy",
    "jupyter",
    "notebook",
    "ipykernel",
    "tqdm"
]

for lib in bibliotecas:
    try:
        versao = pkg_resources.get_distribution(lib).version
        print(f"{lib}: {versao}")
    except pkg_resources.DistributionNotFound:
        print(f"{lib}: não instalada")

# ------------------------------------------------------------
# 7. Lista completa de pacotes instalados
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("LISTA COMPLETA DE PACOTES INSTALADOS")
print("=" * 80)

try:
    resultado = subprocess.run(
        [sys.executable, "-m", "pip", "freeze"],
        capture_output=True,
        text=True
    )

    print(resultado.stdout)

except Exception as e:
    print("Não foi possível listar os pacotes instalados.")
    print(f"Erro: {e}")

# ------------------------------------------------------------
# 8. Salvar relatório em arquivo txt
# ------------------------------------------------------------

nome_arquivo = "relatorio_ambiente_computacional.txt"

with open(nome_arquivo, "w", encoding="utf-8") as f:
    f.write("RELATÓRIO DO AMBIENTE COMPUTACIONAL\n")
    f.write("=" * 80 + "\n\n")

    f.write("INFORMAÇÕES DO PYTHON\n")
    f.write("=" * 80 + "\n")
    f.write(f"Versão do Python: {sys.version}\n")
    f.write(f"Executável Python: {sys.executable}\n")
    f.write(f"Ambiente virtual: {os.environ.get('VIRTUAL_ENV', 'Não identificado')}\n\n")

    f.write("INFORMAÇÕES DO SISTEMA OPERACIONAL\n")
    f.write("=" * 80 + "\n")
    f.write(f"Sistema operacional: {platform.system()}\n")
    f.write(f"Versão do sistema: {platform.version()}\n")
    f.write(f"Release: {platform.release()}\n")
    f.write(f"Arquitetura: {platform.architecture()[0]}\n")
    f.write(f"Nome da máquina: {platform.node()}\n")
    f.write(f"Processador: {platform.processor()}\n\n")

    try:
        import psutil
        memoria = psutil.virtual_memory()
        f.write("CPU E MEMÓRIA RAM\n")
        f.write("=" * 80 + "\n")
        f.write(f"Núcleos físicos da CPU: {psutil.cpu_count(logical=False)}\n")
        f.write(f"Núcleos lógicos da CPU: {psutil.cpu_count(logical=True)}\n")
        f.write(f"Memória RAM total: {memoria.total / (1024 ** 3):.2f} GB\n\n")
    except ImportError:
        f.write("CPU E MEMÓRIA RAM\n")
        f.write("=" * 80 + "\n")
        f.write("Biblioteca psutil não instalada.\n\n")

    f.write("VERSÕES DAS PRINCIPAIS BIBLIOTECAS\n")
    f.write("=" * 80 + "\n")
    for lib in bibliotecas:
        try:
            versao = pkg_resources.get_distribution(lib).version
            f.write(f"{lib}: {versao}\n")
        except pkg_resources.DistributionNotFound:
            f.write(f"{lib}: não instalada\n")

    f.write("\nLISTA COMPLETA DE PACOTES INSTALADOS\n")
    f.write("=" * 80 + "\n")
    try:
        resultado = subprocess.run(
            [sys.executable, "-m", "pip", "freeze"],
            capture_output=True,
            text=True
        )
        f.write(resultado.stdout)
    except Exception as e:
        f.write(f"Erro ao listar pacotes: {e}\n")

print("\n" + "=" * 80)
print("RELATÓRIO SALVO")
print("=" * 80)
print(f"Arquivo gerado: {nome_arquivo}")
```

```python
import numpy as np

perplexity_inicio = 5
perplexity_fim = 300
perplexity_num = 15

perplexidades = np.geomspace(
    perplexity_inicio,
    perplexity_fim,
    num=perplexity_num
)

perplexidades = np.round(perplexidades).astype(int)
perplexidades = np.unique(perplexidades)

print(perplexidades)
```
